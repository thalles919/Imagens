```python
import math

comprimento = float(input("Comprimento do cômodo (m): "))
largura = float(input("Largura do cômodo (m): "))
lado_piso = float(input("Lado do piso quadrado (m), ex: 0.60: "))

area = comprimento * largura
area_piso = lado_piso * lado_piso

pisos = area / area_piso
pisos_com_perda = pisos * 1.10
total = math.ceil(pisos_com_perda)

print("Área do cômodo:", area, "m²")
print("Pisos necessários (com 10% de perda):", total)
```
